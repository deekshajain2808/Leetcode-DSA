```python
import csv
search_name = "Riya"
found = False
with open("student.csv","r",newline="") as file:
  reader = csv.DictReader(file)
  next(reader,None)
  for row in reader:
    if row["Name"].lower == search_name.lower:
      print("Roll No:",row["Roll No"])
      print("Name:",row["Name"])
      print("Marks:",row["Marks"])
      found = True
      break
if not found:
     print("Student not found")
```

```python
import csv
total = 0
count = 0
with open("student.csv","r",newline="") as file:
  reader = csv.DictReader(file)
  next(reader,None)
  for row in reader:
    total += int(row["marks"])
    count += 1
if count > 0:
    average = total/count
    print("Average marks:",average)
else:
  print("No student records found")
```

```python
import csv
topper = None
with open("student.csv","r",newline="") as file:
  reader = csv.DictReader(file)
  for row in reader:
    if topper is None or int(row["marks"]) > int(topper["marks"]):
      topper = row
if topper:

  print("Topper",topper["name"])
  print("Highest Marks:",topper["marks"])
else:
  print("No student records found")
```

```python
import csv
with open("student.csv","r",newline="") as source:
     open("toppers.csv","w",newline="") as destination:
      reader = csv.DictReader(source)
      writer = csv.DictWriter(destination,fieldnames=["RollNO","Name","marks"])
      writer.writeheader()
      for row in reader:
        if int(row["marks"]) > 80:
          writer.writerow(row)
print("Topper records saved to toppers.csv")
```
